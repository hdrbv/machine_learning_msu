&\implies \frac{\partial \text{Loss}(\vec{\mathbf{w}})}{\partial w_j} = - \frac{2}{n} \sum\limits_{i=1}^n \underbrace{(y_i - X_i \vec{\mathbf{w}})}_{e_i} \cdot x_{ij} + 2 \lambda w_j\\
&\implies \nabla \text{Loss}(\vec{\mathbf{w}}) = - \frac{2}{n} X^{\top} \vec{e} + 2 \lambda \vec{\mathbf{w}}
\end{align*}

$$
\vec{\mathbf{w}}^*_{\text{ridge}} = (X^{\top}X + n\lambda I)^{-1} X^{\top} y
